# Load original dataset

```python
import json
# 指定目录路径

def load_data(d_name, tag):
    entire_data_list = []
    file_path = f'{d_name}/{d_name}_{tag}_truncated.json'
    with open(file_path, 'r') as f:
        data = json.load(f)
    entire_data_list += data
    print(len(entire_data_list))
    source_list = defaultdict(int)
    for _ in entire_data_list:
        source_list[_['source']] += 1
    print(source_list)
    return entire_data_list
```

```python
import zlib
def get_compression_ratio(input_data):
    data_str = str(input_data).encode('utf-8')
    compressed_data = zlib.compress(data_str, level=9)
    compressed_ratio = len(data_str) / len(compressed_data)
    return compressed_ratio
```

```python
def get_seq_len(data):
    seq_len_list = []
    for line in data:
        seq_len = 0
        for _ in line['conversations']:
            seq_len += len(_['value'])
        seq_len_list.append(seq_len)
    return sum(seq_len_list) / len(seq_len_list)
```

```python
def statistics(d_name, tag):
    print(f"=============={d_name} - {tag} ===============")
    a = load_data(d_name, tag)
    a_id = set([_['id'] for _ in a])
    print("total unique id", len(a_id))
    turn_num_a = [len(_['conversations']) for _ in a]
    print("Avg turn length", sum(turn_num_a) / len(turn_num_a))
    print("Avg seq length", get_seq_len(a))
    # print("Compression ratio is ", get_compression_ratio(a))

```

```python

statistics('zip', '10000')
statistics('token_mistral_re', 'random')
statistics('token_mistral_re', 'diversity')
statistics('token_mistral_re', 'perplexity')
statistics('token_mistral_re', 'deita')
statistics('token_mistral_re', 'IFD')
```
